```python
#11. Prime Number Check
#for loop
n = 7
count =0
for i in range(1,n+1):
  if n%i==0:
    count = count + 1
if count == 2:
  print('prime')
else:
  print('not a prime')

#while loop
n = 7
i = 1
count = 0
while i<=7:
  if n%i==0:
    count = count + 1
  i = i + 1
if count == 2:
  print('prime')
else:
  print('not prime')
#function
def prime_num(n):
  count = 0

  for i in range(1,n+1):
    if n%i==0:
      count = count +1

  if count == 2:
    print('prime')
  else:
    print('not prime')
prime_num(7)
```

```python
#12. Even Numbers from m to n
#for loop
m = 3
n = 10
for i in range(m,n+1):
  if i %2==0:
    print(i)
#while loop
m = 3
n = 10
while m<=n:
  if m%2==0:
    print(m)
  m = m +1
 #funtions
def even_num(m,n):
  for i in range(m,n+1):
    if i % 2==0:
      print(i)
even_num(3,10)

```

```python
#13. Odd Numbers from m to n
#for loop
m =3
n =10
for i in range(m,n+1):
  if i%2!=0:
    print(i,end='  ')
#while loop
m = 3
n = 10
while m<=n:
  if m%2!=0:
    print(m,end='  ')
  m = m + 1
#functions
def odd_num(m,n):
  for i in range(m,n+1):
    if i % 2==0:
      print(i,end='  ')
odd_num(3,10)
```

```python
#14. Count of Even and Odd Numbers
#for loop
m = 3
n = 7
count_even =0
count_odd = 0

for i in range(m,n+1):
  if i%2==0:
    count_even =count_even + 1
  elif i%2!=0:
    count_odd =count_odd + 1
print('even=',count_even)
print('even=',count_odd)

#while loop
m = 3
n = 7
count_even = 0
count_odd = 0
while m<=n:
  if m%2==0:
    count_even = count_even + 1
  elif m%2!=0:
    count_odd = count_odd + 1
  m = m + 1
print('even=',count_even)
print('odd=',count_odd)

#functions
def count_e_o(m,n):
  count_even =0
  count_odd=0

  for i in range(m,n+1):
    if i%2==0:
      count_even = count_even + 1
    elif i%2!=0:
      count_odd = count_odd + 1
  print('even=',count_even)
  print('odd=',count_odd)

count_e_o(3,7)
```

```python
#15. Reverse a String
#for loop
s ='hello'
rev = ' '
for i in s:
  rev = i + rev
print(rev)

#while loop
s ='hello'
rev =' '
i = len(s) -1
while i>=0:
  rev = rev + s[i]
  i = i - 1
print(rev)
```

```python
#16. Check for Palindrome String
#for loop
s ='madam'
rev =''
for i in s:
  rev = i + rev

if s == rev:
  print('palindrome')
else:
  print('not palindrome')
#while loop
s ='madam'
i =len(s) -1
rev =''
while i>=0:
  rev = rev +s[i]
  i = i - 1

if s == rev:
  print('palindrome')
else:
  print('not pa')

#functions
def palindrome_string(s):
  rev =s[: : -1]

if s == rev:
  print('palindrome')
else:
  print('not a palindrome')


palindrome_string('madam')
```

```python
#17. Sum of Digits
#for loop
n = 123
total = 0

for i in str(n):
    total =total + int(i)

print(total)
#while loop
n = 123
total = 0

while n > 0:
    digit = n % 10
    total += digit
    n = n // 10

print(total)
```

```python
#product of digits
#for loop
n = 123
total = 1
for i in str(n):
  total = total * int(i)
print(total)

#while loop
n = 123
product = 1

while n > 0:
    digit = n % 10
    product *= digit
    n = n // 10

print(product)
```
