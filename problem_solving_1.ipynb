```python
#1. Print Numbers from 1 to n
#for loop
n = 5
for i in range(1,n+1):
  print(i)
#while loop
n =5
i =1
while i<=n:
  print(i)
  i = i +1
#functions
def numbers(n):
  for i in range(1,n+1):
    print(i)
numbers(10)
```

```python
#3.print numbers from n to 1 in reverse
#for loop
n = 5
for i in range(n,0,-1):
  print(i)
#while loop
n = 5
i = 1
while n>=i:
  print(n)
  n = n - 1
#functions
def reverse(n):
  for i in range(n,0,-1):
    print(i)
reverse(5)
```

```python
#5. Sum of n Natural Numbers
#for loop
n = 5
total = 0
for i in range(1,n+1):
  total =total + i
print(total)
#while loop
n = 5
i = 1
total = 0
while i<=n:
  total = total + i
  i = i + 1
print(total)

#functions
def sum(n):
  total = 0

  for i in range(1,n+1):
    total = total + i
  return total
sum(5)
```

```python
#6.factorial of a number
#for loop
n = 5
factorial = 1
for i in range(1,n+1):
  factorial = factorial * i
print(factorial)
# while loop
n =5
i = 1
factorial = 1
while i<=n:
  factorial = factorial * i
  i = i + 1
print(factorial)
#functions
def factorial(n):
  fact = 1

  for i in range(1,n+1):
    fact = fact * i
  return fact
print(factorial(5))
```

```python
#8.product of m to n numbers
#for loop
m = 2
n = 4
total =1
for i in range(m,n+1):
  total = total * i
print(total)
#while loop
m = 2
n = 4
total = 1
while m<=n:
  total = total * m
  m = m + 1
print(total)
#functions
def product(m,n):
  total = 1

  for i in range(m,n+1):
    total = total * i
  return total
print(product(2,4))



```
